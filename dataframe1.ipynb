```python
import pandas as pd

# 이름, 나이, 성별 데이터로 DataFrame 생성
df = pd.DataFrame({
    '이름': ['철수', '영희', '민수', '지영', '준호'],
    '나이': [20, 21, 20, 22, 21],
    '성별': ['남', '여', '남', '여', '남']
})

print(df)
```

```python
# 1) info() 메서드
# 행 개수, 열 개수, 열 이름, 결측값이 아닌 데이터 개수, 데이터 타입 확인

print("===== info =====")
# DataFrame의 전체 정보를 출력
df.info()
```

```python
# 2) shape, size, ndim
# 행과 열의 크기

print("shape:", df.shape) # 출력 (행 개수, 열 개수)

print("행 개수:", df.shape[0])
print("열 개수:", df.shape[1])

print(df.shape, df.size, df.ndim)
```

```python
# 3) dtypes
# 각 열의 데이터 타입

print("===== dtypes =====")
print(df.dtypes)

# 특정 열의 데이터 타입 확인
print(df['나이'].dtype)
print(df['이름'].dtype)
```

```python
# 4) columns
# 열 이름 확인

print("===== columns =====")
print(df.columns)

# 이름만 확인
print(df.columns.tolist())

# in 연산자를 사용하여 특정 열의 존재 유무 확인
print('나이' in df.columns)
print('주소' in df.columns)
```

```python
# 5) index
# 행 레이블 인덱스 확인(위치값 인덱스 아님)
print("===== index =====")
print(df.index)
```

```python
# 6) head, tail
print("===== head =====")
print(df.head()) # 기본값으로 5행
print(df.head(3))

print("===== tail =====")
print(df.tail())
print(df.tail(3))
```

```python
# 7) unique, nunique
# 고유값 확인, 고유값 개수
print(df['성별'].unique())
print(df['나이'].unique())

# 고유값 개수
print(df['성별'].nunique())
```

12) 판다스 DataFrame타입 데이터 추출하기

- 행과 열을 동시에 선택 <br>
loc[]: 인덱스 → 레이블 기준으로 행/열 선택 <br>
iloc[]: 인덱스 → 숫자 위치 기준으로 행/열 선택 <br>

- 행과 열 중 하나만 선택 <br>
df[]: <br>
연산자 → 열이름 → 열을 선택하면 → 행은 전체 → 특정 열의 모든 행 <br>
조건식 → 행 선택 : 행을 선택하면 → 열은 전체 → 특정 행의 모든 열 <br>
슬라이싱 → 행 선택 : 행을 선택하면 → 열은 전체 → 특정 행의 모든 열

```python
# 실습용 데이터 프레임
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88]
}, index=['A', 'B', 'C', 'D', 'E'])

print(df)
```

- 행 레이블 인덱스 → A, B, C, D, E

- 행 위치 → 0, 1, 2, 3, 4



- 열 이름 → 이름, 학년, 점수

- 열 위치 → 0, 1, 2

1) loc인덱스 : 행 인덱스, 열이름 사용

```python
# loc[행 인덱스, 열 이름]

print(df.loc['B', '점수']) # 92

# 여러행, 여러 열 선택 시 []안에 리스트 사용
print(df.loc[['B', 'D'], ['이름', '점수']])
```

```python
# loc[행 인덱스 슬라이싱, 열 이름 슬라이싱]
df.loc['B':'D', '이름':'학년']
```

2) iloc인덱스 : 행 위치, 열 위치 사용

```python
# iloc[행 위치, 열 위치]

print(df.iloc[1, 2]) # 92

# 여러행, 여려열 선택
print(df.iloc[[1, 3], [0, 2]])
```

```python
# iloc[행 위치 슬라이싱, 열 위치 슬라이싱]
df.iloc[1:4, 0:2]
```

2. [] 인덱싱 연산자
- [값] -> 열선택
- [식] -> 행선택

```python
# 1) 열 선택
# - df[숫자]
# - "열 이름이 0인 열"

# #0번째 열 아님. 0번째 열은 iloc[]인덱스를 사용하여야 한다 
# df[0] # error

# - df['이름']

# 열 이름 위치
# 두개이상의 열은 []안에 리스트를 사용하여 []안에 하나의 값형태로 입력
# df[['이름', '점수']]
# df["이름", "점수"] error! []연산자 안에는 하나의 표현식만이 올 수 있다
# - []연산자안에 매개값으로 리스트 타입이 입력되면 반환되는 타입은 하나 이상의 열을 반환하므로 DataFrame타입을 반환한다. []연산자안에 스칼라 열이 입력되면 반환타입은 Series타입을 반환.
```

```python
df['점수'] # Series
```

```python
df[['이름', '점수']] # DataFrame
```

```python
# 하나의 열이지만 리스트형태로 입력되면 반환타입은 1열짜리 DataFram타입이 반환된다.
df[['점수']] # DataFrame
```

2) 행 선택

```python
# 1) df[슬라이싱] : []연산자안에 슬라이싱 표현식
df[1:4] # 행을 위치 기준으로 위치 1부터 3까지 선택
```

```python
# 2) df[열조건] : []연산자안에 열조건 표현식
df[df['점수'] >= 90] # 조건에 맞는 행 선택
```

```python
# 위랑 같음
temp = df['점수']
df[temp >= 90]
```

- Python 리스트는 > 연산자를 리스트 전체에 적용하면 원소별 비교를 하지 않는다. 따라서 리스트 컴프리헨션이나 반복문을 사용해야 한다.
- pandas Series는 내부적으로 NumPy의 배열 연산 등을 활용하여 > 같은 연산자를 원소별(element-wise)로 적용할 수 있다.​

1. 입력, 수정, 삭제

(간단한 CRUD방법만 실습하고, 데이터 전처리 챕터에서 좀 더 상세히 설명한다)

```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
# 새로운 열 추가
df['성적'] = ['B', 'A', 'C', 'A+', 'B+']
print(df)
```

```python
# 계산해서 새로운 열 만들기

df['총점'] = df['점수'] + df['출석']
print(df)
```

```python
# 특정 값 수정

df.loc[0, '점수'] = 90
print(df)
```

```python
df.loc[df['점수'] < 80, '점수'] = 80
print(df)
```

삭제 drop(열이름 or 레이블 인덱스)메서드

* drop()에 매개값으로 위치값을 사용할 수 없다
- 행삭제 : drop('행 레이블', axis=0)
- axis=0이면 첫번째 매개값을 행 레이블(인덱스)에서 찾음.
- axis=1이면 첫전째 매개값을 열 레이블(이름)에서 찾음.
- axis=0은 행축(세로방향으로 삭제 - 행들을 세로방향? 하나이상의 행들들을 삭제 가능)

```python
df = df.drop(0, axis=0) # axis는 생략하면 0
```

```python
# '총점'을 열 방향(axis=1)에서 삭제
df.drop('총점', axis=1) # axis는 생략하면 0
```

```python
print(df.shape) # (행, 열)
print(df.columns) # 가로축 레이블
print(df.index) # 세로축 레이블
print(df.dtypes) # 데이터 타입
```

```python
# 2. 통계 관련 메서드
# 1) 전체 통계량

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df.count())

print(df.sum(numeric_only=True))

print(df.mean(numeric_only=True))

print(df.median(numeric_only=True))

print(df.min(numeric_only=True))

print(df.max(numeric_only=True))

print(df.std(numeric_only=True))

print(df.std(numeric_only=True))

print(df['점수'].max() - df['점수'].min())
```

```python
print(df.describe())
```

```python
# 2) 특정 열의 통계량

print(df['점수'].mean())
print(df['점수'].median())
print(df['점수'].min())
print(df['점수'].max())
print(df['점수'].std())
```

```python
# 수치형열의 describe() vs 범주형열의 describe()

# 3) 기타

print(df.isnull())
# True  → 결측값
# False → 결측값 아님

print(df.isnull().sum()) # 결측값 개수

print(df.duplicated()) # 중복 행 여부
# True  → 중복 행
# False → 중복 아님

print(df['학년'].unique()) # 고유값 확인 : 학년 열에 어떤 값들이 존재하는지 확인한다.
print(df['학년'].nunique()) # 고유값 개수

print(df['학년'].value_counts()) # 값별 개수

print(df['점수'].isnull()) # 특정 열의 결측값
```

```python
# 학년의 데이터 타입을 category 타입(범주형)으로 변환한 결과를 
# 다시 df['학년']에 저장하여 df의 학년 데이터 타입을 변경
df['학년'] = df['학년'].astype('category')

print(df['학년'].dtype)
```
