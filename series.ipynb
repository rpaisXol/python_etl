1) 시리즈의 인덱스 <br>
파이썬 랭기지의 인덱스의 개념은 위치개념 이지만, 판다스의 데이터타입의 인덱스 개념은 위치값인덱스와 레이블인덱스를 구분한다. <br>
판다스에서 인덱스라 함은 "레이블 인덱스"를 의미한다. 레이블 인덱스를 지정하지 않으면 자동으로 0, 1, 2, ...가 부여된다.

```python
import pandas as pd

s1 = pd.Series([10, 20, 30, 40, 50])
print(s1)
print(type(s1))
```

```python
# 인덱스와 위치 번호가 우연히 똑같은 경우
print(s1.loc[0])    # 10 → 인덱스 0
print(s1.iloc[0])   # 10 → 0번째 위치

# Series는 항상 인덱스를 가진다.
# 인덱스를 지정하지 않으면 pandas가 0, 1, 2, ...를 자동으로 만든다.
# 이때 인덱스와 위치 번호가 같아 보일 뿐, 개념적으로는 다르다.
```

```python
# 인덱스 직접 지정
s2 = pd.Series(
    [90, 80, 70],
    index=["국어", "영어", "수학"]
)
print(s2)
```

```python
print(s2.loc["국어"])    # 인덱스 이름 "국어" → 90
print(s2.iloc[0])        # 0번째 위치 → 90

# 아래 코드는 에러
# print(s2.loc[0])  
```

2) 데이터프레임의 인덱스

```python
import pandas as pd

df1 = pd.DataFrame({
    "이름": ["홍길동", "김철수", "이영희"],
    "나이": [20, 30, 25],
    "점수": [90, 80, 95]
})

# 행 위치번호와 행 인덱스가 동일
print(df1.loc[0])
print(df1.iloc[0])


df2 = pd.DataFrame({
    "이름": ["홍길동", "김철수", "이영희"],
    "나이": [20, 30, 25],
    "점수": [90, 80, 95]
}, index=["a", "b", "c"])

print(df2)

print(df2.loc["a"])
print(df1.iloc[0])

# error
# print(df2.loc[0])
```

- 데이터 분석 코드에서는 레이블 인덱스 메서드인 loc[]는 잘 사용되지 않는다. 
- 대부분은 경우 위치값 메서드인 iloc[]를 사용한다.
- 간단하게 데이터플레임에서 loc, iloc를 사용하여 행을 추출하는 방법을 배웠다.
- 데이터플레임에서 행, 열을 추출하는 다양한 방법들은 뒤에서 자세히 다룬다.

2. 슬라이싱과 산술연산

```python
import pandas as pd

# Series 생성
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]


# loc[] : 인덱스 기준
# vs 
# iloc[] : 위치값(번호) 기준
print(temperature.loc["Jan":"Mar"]) # Mar까지 포함 -> Jan, Feb, Mar
print(temperature.iloc[0:2]) # 2는 포함이 안됨 -> 0, 1
```

- 데이터 전처리 과정에서는 위치 기반 작업이 필요한 경우가 많다.  iloc[]메서드를 사용하는 것이 편하다.
- 특별한 경우 위치가 아닌 의미 있는 이름으로 데이터를 선택해야 한다면 loc[]를 사용하자.

다음의 경우 iloc[]가 편한다. <br>
- 첫 번째/마지막 데이터 선택
- 특정 위치의 데이터 선택 
- 데이터 샘플 추출 
- 반복적인 위치 기반 처리 
- 인덱스 이름이 변경될 가능성이 있는 데이터

```python
# 3. Series 기술통계 메서드
import pandas as pd

# Series 생성
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]


# 슬라이싱
print(temperature.iloc[:3]) # 앞에서 3개
print(temperature.iloc[-3:]) # 뒤에서 3개
print(temperature.iloc[1:5]) # 2번째~5번째
print(temperature.iloc[::2]) # 홀수 번째
print(temperature.iloc[::-1]) # 역순

# 조건 슬라이싱
print(temperature[temperature >= 20])
print(temperature[temperature < 0])

# 산술연산
print(temperature + 1)
print(temperature * 2)
print(temperature - 5)
print(temperature / 2)
print(temperature.iloc[:3] + 10)
```

3. Series 기술통계 메서드

```python
import pandas as pd

# Series 생성
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

print("개수: ", temperature.count())
print("합: ", temperature.sum())
print("최대값: ", temperature.max())
print("최소값: ", temperature.min())

print("평균: ", temperature.mean())
print("중앙값: ", temperature.median())
# 12개 중복된 값이 없어서 12개 공동1위로 모두 나옴
print("최빈값: ", temperature.mode()) # 범주형 타입의 대표값으로 사용

print("분산: ", temperature.var()) # 각 데이터에서 평균을 뺀 후 제곱 → 모두 더함 → 데이터 개수로 나눔
print("표준편차: ", temperature.std()) # 분산에 루트

# 기술통계를 한 번에 확인
print(temperature.describe())
```

4. Series 값 수정, 삭제, 추가

```python
import pandas as pd

# Series 생성
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

temperature.iloc[0:3] = [1, 2, 3]
print(temperature)

# drop메서드는 원본을 삭제하는 것은 아니다.
# 위치값 사용
temperature2 = temperature.drop(temperature.index[0])
print(temperature2)

# 위치값이 아닌 레이블 인덱스 사용
temperature3 = temperature.drop("Jan")
print(temperature3)

print(temperature)

# 값 추가
temperature.loc["Avg"] = temperature.mean()
print(temperature)
```

5. 범주형 데이터를 가진 시리즈

```python
# 5. 범주형 데이터를 가진 시리즈
import pandas as pd

gender = pd.Series(
    ["남", "여", "여", "남", "남", "여", "여", "남", "여"],
    dtype="category"
)

print(gender)
print(gender.dtype)
print(gender.cat.categories) # 범주의 종류 확인 -> "남", "여" 확인 
print(gender.nunique()) # 범주의 개수
print(gender.value_counts()) # 범주별 데이터 개수
print("최빈값: ", gender.mode()) # 최빈값 : 범주형의 대표값은 최빈값. 평균, 중앙값등은 의미가 없다
print(gender.cat.codes) # 판다스가 내부적으로 관리하는 범주의 인코딩 값

print(gender.describe())

# 범주 추가
gender = gender.cat.add_categories(["기타"])

print(gender.cat.categories)
print(gender)
```

10) 판다스 Series 실습


```python
import pandas as pd

# 데이터 준비
score = pd.Series(
    [78, 85, 82, 90, 88, 92],
    index=["1학년중간", "1학년기말",
           "2학년중간", "2학년기말",
           "3학년중간", "3학년기말"]
)

print(score)
```

```python
# 문제 1. Series 기본 정보
# 다음 정보를 출력하라.
# 데이터 개수
# shape
# 데이터 타입
# 인덱스
# 값

print("데이터 개수:", score.count())
print("shape:", score.shape)
print("데이터 타입:", score.dtype)
print("인덱스:", score.index)
print("값:", score.values)
```

```python
# 문제 2. 학년별 성적 조회
# loc를 이용하여 다음을 출력하라.
# 1학년 중간고사 성적
# 2학년 기말고사 성적
# 3학년 중간고사 성적

print("1학년 중간고사 성적:", score.loc["1학년중간"])
print("2학년 기말고사 성적:", score.loc["2학년기말"])
print("3학년 중간고사 성적:", score.loc["3학년중간"])
```

```python
# 문제 3. iloc를 이용한 조회
# iloc를 이용하여 다음을 출력하라.
# 첫 번째 시험 성적
# 마지막 시험 성적
# 1학년 두 시험 성적
# 3학년 두 시험 성적

print("첫 번째 시험 성적:", score.iloc[0])
print("마지막 시험 성적:", score.iloc[-1])
print("1학년 두 시험 성적:", score.iloc[0:2])
print("3학년 두 시험 성적:", score.iloc[4:6])
```

```python
# 문제 4. 성적 슬라이싱
# iloc를 이용하여 다음을 출력하라.
# 처음 3개의 시험 성적
# 마지막 3개의 시험 성적
# 짝수 번째 위치의 성적
# 역순으로 성적 출력

print("처음 3개의 시험 성적:", score.iloc[:3])
print("마지막 3개의 시험 성적:", score.iloc[-3:])
print("짝수 번째 위치의 성적:", score.iloc[::2])
print("역순으로 성적 출력:", score.iloc[::-1])
```

```python
# 문제 5. 기술통계
# 다음 값을 구하라.
# 평균
# 최댓값
# 최솟값
# 중앙값
# 표준편차
# 분산
# 합계
# 그리고 describe()를 이용하여 전체 기술통계량을 출력하라.


print("평균:", score.mean())
print("최댓값:", score.max())
print("최솟값:", score.min())
print("중앙값:", score.median())
print("표준편차:", score.std())
print("분산:", score.var())
print("합계:", score.sum())
```

```python
# 문제 6. 조건 검색
# 다음 조건에 해당하는 시험을 출력하라.
# 90점 이상
# 80점 이상 90점 미만
# 평균보다 높은 성적
# 70점 미만인 성적


print("90점 이상:", score[score >= 90])
print("80점 이상 90점 미만:", score[(score >= 80) & (score < 90)])
print("평균보다 높은 성적:", score[score > score.mean()])
print("70점 미만인 성적:", score[score < 70])
```

```python
# 문제 7. 성적 수정·추가
# 다음 작업을 수행하라.
# 1학년중간 성적을 80점으로 수정
# 3학년기말 성적을 95점으로 수정
# 재수강이라는 인덱스를 추가하고 성적 88점을 입력
# 재수강 데이터를 삭제

score.loc["1학년중간"] = 80
score.loc["3학년기말"] = 95
score.loc["재수강"] = 88
del score["재수강"]

print(score)
```

```python
# 문제 8. 범주형 데이터 확인
# 다음 정보를 출력하라.
# print(grade.dtype)
# print(grade.cat.categories)
# print(grade.cat.codes)
# 그리고 일반적인 문자열 Series와 범주형 Series의 차이를 확인하라.

print(grade.dtype)
print(grade.cat.categories)
print(grade.cat.codes)
```

```python
# 문제 9. 범주별 개수 확인
# 각 학점이 몇 번 나타나는지 구하라.
print(grade.value_counts())
```

```python
# 문제 10. 범주형 데이터 수정
# 다음 작업을 수행하라.
# 1학년중간의 등급을 A로 변경
# 2학년중간의 등급을 C로 변경
# C라는 범주가 없을 경우 범주를 먼저 추가
# 전체 학점 분포를 다시 출력
grade.loc["1학년중간"] = "A"
grade = grade.cat.add_categories(["C"])
grade.loc["2학년중간"] = "C"
print(grade.value_counts())
```
