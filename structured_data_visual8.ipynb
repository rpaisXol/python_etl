1. concat() 함수
- 전처리 - 데이터 결합하기
- concat() 함수 : 같은 구조의 데이터를 합치는 것
- axis=0 → 행을 아래로 붙인다
- axis=1 → 열을 옆으로 붙인다
- SQL의 JOIN과는 다르다.
- concat은 단순히 축을 기준으로 이어 붙이는 것이고, 
- 공통 키를 기준으로(SQL의 JOIN 유사) 데이터를 결합하려면 merge()나 join()을 사용한다.

```python
# import pandas as pd

# pd.concat( 
#     objs, # 연결할 데이터프레임들의 리스트
#     axis=0, # 연결 축 (0: 행 방향, 1: 열 방향)
#     join="outer", # 결합 방식 ("outer", "inner")
#     ignore_index=False, # 인덱스를 무시하고 새로 생성할지 여부
#     keys=None # 결과에 키 추가할지 여부
# )
```

join은 “공통 부분만 붙일 것인가, 전체를 유지할 것인가”를 결정하는 옵션이다. SQL문의 JOIN같은 것이 아니다.

```python
import pandas as pd

df1 = pd.DataFrame({
    "이름": ["홍길동", "김철수"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["이영희", "박민수"],
    "점수": [85, 95]
})

display(df1)
display(df2)
```

ignore_index=False : 기본값

```python
# axis=0 위아래 결합(기본값) -> 행추가
df = pd.concat([df1, df2], ignore_index=False, axis=0) # ignore_index 기본값 False
display(df)
```

ignore_index=True 

```python
df = pd.concat([df1, df2], ignore_index=True, axis=0)
display(df)
```

axis=1 : 생략하면 기본값 0

```python
# axis=0 좌우 결합 -> 열추가
df = pd.concat([df1, df2], axis=1) # axis=0 행축, axix=1 열축 기본값 0
display(df)
```

- join="outer" vs join="inner"
- join은 “공통 부분만 붙일 것인가, 전체를 유지할 것인가”를 결정

```python
df1 = pd.DataFrame({
    "이름": ["김민수", "이서연"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["박지훈", "최유진"],
    "등급": ["A", "B"]
})
display(df1)
display(df2)
```

두 데이터 프레임의 공통된 열은 "이름"열만 존재한다.

```python
df = pd.concat([df1, df2], join="outer", ignore_index=True) # join생략하면 기본값이 outer
display(df)
```

```python
df = pd.concat([df1, df2], join="inner", ignore_index=True) # inner join -> 공통 컬럼만 남김
display(df)
```

2. merge() 함수

- merge()는 두 DataFrame의 공통된 열(키)을 기준으로 데이터를 결합할 때 사용하는 함수
- SQL의 JOIN문의 경험이 있다면 merge()는 이해하기 쉽다.(JOIN 개념 자체는 동일하다)

```python
import pandas as pd

df_customer = pd.DataFrame({
    "고객ID": [101, 102, 103, 104],
    "이름": ["홍길동", "김철수", "이영희", "박민수"]
})

df_score = pd.DataFrame({
    "고객ID": [101, 102, 103, 105],
    "점수": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)

result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="inner" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="left"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="right"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="outer"
)

display(result)
```

- 판다스 라이브러리에는 merge() 함수와 유사한 DataFrame.merge(), DataFrame.join() 같은 메서드들도 있다. 
- 비슷한 기능을 가진 merge 메서드와 join 메서드가 존재하는 이유는 사용 상황과 코드 표현 방식이 조금 다르기 때문이다. 우선은 merge()함수으로도 충분하다.

29) DataFrame 다루기 - 결측치(누락값) 처리 1

- 결측치(누락값) 메서드
1. isna(): 누락값 위치 확인
2. isna().sum(): 열별 누락값 개수
3. dropna(): 누락값이 있는 행 삭제
4. fillna(): 누락값을 다른 값으로 대체
5. mean(): 기본적으로 누락값 대체
6. sum(): 기본적으로 누락값 제외
7. count(): 누락값 제외한 데이터 개수

※ 삭제 및 대체 판단 <br>
- 결측치를 억지로 채워 넣으면 오히려 분석 모델이 추측에 의존하게 되어 결과가 심각하게 왜곡될 수 있으니 주의
- 삭제 권장 (50% 이상): 결측치가 해당 열의 50%를 넘는다면, 어떤 값으로 채워도 신뢰성이 부족해집니다. 
- 이럴 땐 해당 열 전체를 삭제하는 것이 합리적입니다.
- 대체 권장 (20~30%): 결측치가 20~30% 수준이라면 평균, 중앙값, 혹은 회귀 분석을 통한 대체법을 사용하여 데이터를 살리는 것이 좋습니다.
- 행 삭제: 결측치 수가 극소수라면 해당 행만 삭제해도 분석 결과에 큰 무리가 없습니다.

```python
# 누락값 확인 isna() 또는 isnull()을 사용한다.
# None은 python에서 값이 없는 객체
# NaN은 numpy, pandas의 숫자데이터의 결측값(None을 입력해도 NaN으로 처리됨)
import pandas as pd

data = {
    "이름": ["김민준", "이서연", "박지훈", "최유진", "정현우"],
    "점수": [85, None, 90, 75, None],
    "출석": [True, True, None, True, False]
}

df = pd.DataFrame(data)

print(df) # 데이터프레임은 객체타입은 None, 숫자타입은 NaN으로 처리

print(df.isna())
print(df.isnull())
```

※ 불값을 sum연산을 하면 True -> 1, False -> 0 으로 계산된다 <br>
-> sum()메서드를 이용하여 결측치 개수 확인하는 방법 -> 결측치가 3개 있음을 확인

```python
# 하나의 열을 분리 후 시리즈 메서드 isna()를 사용해도 된다.
print(df["점수"].isna()) # 점수열의 시리즈메서드 isna() 
```

조건식 사용

```python
cond = df["점수"].isna() # [False, True, False, False, True] 시리즈 반환
print(df[cond]) # df[[False, True, False, False, True]] => True인 인덱스1, 4행 출력

# 한줄의 코드로 만들면
print(df[df["점수"].isna()]) # df[조건] -> 행반환

# cond가 True 이름만 출력 => loc[] 인덱스 사용
print(df.loc[cond, "이름"]) # df[조건] -> 행반환
print(df.loc[df["점수"].isna(), "이름"]) # df[조건] -> 행반환 
```

```python
# 누락값 개수 확인
df.isna().sum()
# 이름    0 → 누락값 0개
# 점수    2 → 누락값 2개
# 출석    1 → 누락값 1개
```

```python
print(df.isna().any()) # 데이터프레임 열에 누락값이 하나라도 있으면 True
# 이름    False
# 점수     True
# 출석     True

print(df.isna().all()) # 데이터프레임 열이 모두 누락값이면 True
# 이름    False
# 점수    False
# 출석    False
```

```python
# 누락값 처리
# 누락값 처리는 크게 삭제와 대체 → 삭제보다는 대체

# dropna() 누락값이 있는 행 삭제한 데이터프레임을 반환
df2 = df.dropna()
print(df2)

# 원본의 행은 삭제 안됨
print(df)
```

```python
# 특정열의 누락값이 있는 행 삭제한 데이터프레임 반환
df2 = df.dropna(subset=["점수", "출석"])
print(df2)

df2 = df.dropna(subset=["점수"])
print(df2)

# subset=df.columns 기본값
```

```python
# 선택열의 누락값을 특정 값으로 대체 후 데이터셋(시리즈, 데이터프레임)반환 → fillna()
print(df["점수"].fillna(0)) # 열.fillna() 시리즈메서드 fillna() → 시리즈 반환
print(df[["점수","출석"]].fillna(0)) # 데이터프레임 메서드 fillna() → 데이터프레임 반환

# 원본에 대입
df["점수"] = df["점수"].fillna(0)
df
```

```python
# 범주(Category)의 경우 최빈값 대체

import pandas as pd

data = {
    "이름": ["김민준", "이서연", "박지훈", "최유진", "정현우"],
    "성별": ["M", "F", None, "F", None] 
}

df = pd.DataFrame(data)

# mode()의 반환타입은 시리즈 타입이다. 결과값을 값을 가져오기 위해 위치값 [0]이 필요
mode_gender = df["성별"].mode()[0]

df["성별"] = df["성별"].fillna(mode_gender)

print(df)
```

```python
# 데이터 준비
students = pd.DataFrame({
    "학생ID": [101, 102, 103, 104, 105],
    "이름": ["김민준", "이서연", "박지훈", "최유진", "정현우"]
})

scores = pd.DataFrame({
    "학생ID": [101, 102, 104, 105],
    "점수": [85, None, 90, 75]
})
```

```python
# 학생id를 기준으로 두 DataFrame을 merge()하시오.
result = pd.merge(
    students,
    scores,
    on="학생ID",
    how="left" # left join
)
print(result)
```

```python
# 모든 학생이 결과에 포함
result = pd.merge(
    students,
    scores,
    on="학생ID",
    how="left" # left join
)
print(result)
```

```python
# 점수의 결측치 개수를 확인
missing_scores_count = result["점수"].isna().sum()
print("점수의 결측치 개수:", missing_scores_count)
```

```python
# 점수의 평균값으로 결측치를 대체
mean_score = result["점수"].mean()
result["점수"] = result["점수"].fillna(mean_score)
print(result)
```
