35) DataFrame 다루기 - cut()함수 : 숫자형 -> 구간 범주형

- pandas.cut()은 연속형 숫자 데이터를 구간으로 나누어 범주형 데이터로 변환하는 함수다.

```python
import pandas as pd

scores = pd.Series([45, 55, 65, 75, 85, 95])

result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],       # 구간의 경계값
    labels=["하", "중", "상"]    # 각 구간에 붙일 이름
    # 0 ───── 60 ───── 80 ───── 100
    #     하       중       상
)

display(result)
```

```python
# ​labels 생략하면
result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],
    right=True # 오른쪽 구간 포함 -기본값
)

display(result)
```

```python
# right=False 이면
scores = pd.Series([45, 55, 65, 75, 85, 95])
result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],
    right=False
)

display(result)
```

```python
# cut()함수를 이용하여 범주형 구간 파생변수 추가 예

df = pd.DataFrame({
    "이름": ["철수", "영희", "민수", "지수"],
    "점수": [55, 72, 85, 95]
})

df["등급"] = pd.cut(
    df["점수"],
    bins=[0, 60, 80, 100],
    labels=["하", "중", "상"]
)

display(df)
```

1. str 접근자
- pandas의 .str 접근자는 Series의 문자열을 한꺼번에 처리할 때 사용한다.

2. 정규표현식 사용

- contains()는 찾아서 True/False를 반환
- extract()는 찾은 내용을 반환

예제에 사용되는 정규표현식

숫자가 포함되어 있는지 / 숫자 추출

```python
import pandas as pd

df = pd.DataFrame({
    "코드": ["A100", "B200", "ABC", "C300"]
})

# 숫자가 포함되어 있는지 확인
# \d → 숫자
# + → 1개 이상
df["숫자포함"] = df["코드"].str.contains(r"\d+")

# 숫자 추출
df["숫자"] = df["코드"].str.extract(r"(\d+)")

display(df)
```

영문 대문자가 있는지 / 영문 추출

```python
df = pd.DataFrame({
    "코드": ["A100", "B200", "123", "C300"]
})

df["영문포함"] = df["코드"].str.contains(r"[A-Z]")
df["영문"] = df["코드"].str.extract(r"([A-Z])")

display(df)
```

전화번호 형식 / 전화번호 추출

```python
df = pd.DataFrame({
    "전화번호": [
        "010-1234-5678",
        "02-123-4567",
        "01012345678"
    ]
})

# {3} → 정확히 3개
# {4} → 정확히 4개

df["전화번호형식"] = df["전화번호"].str.contains(
    r"\d{3}-\d{4}-\d{4}"
)

df["전화번호"] = df["전화번호"].str.extract(
    r"(\d{3}-\d{4}-\d{4})"
)

display(df)
```

이메일 주소 / 이메일 도메인 추출

```python
df = pd.DataFrame({
    "이메일": [
        "kim@gmail.com",
        "lee@naver.com",
        "park@daum.net"
    ]
})

# \w+ → 문자/숫자/_가 1개 이상
# \. → 실제 . 문자
# () → extract()에서 추출할 부분

df["이메일여부"] = df["이메일"].str.contains(
    r"@\w+\.\w+"
)

df["도메인"] = df["이메일"].str.extract(
    r"@(\w+\.\w+)"
)

display(df)
```

특정 문자로 시작하는지 / 시작 문자 추출

```python
df = pd.DataFrame({
    "상품코드": ["A100", "A200", "B300", "C400"]
})

df["A로시작"] = df["상품코드"].str.contains(
    r"^A"
)

df["첫문자"] = df["상품코드"].str.extract(
    r"^([A-Z])"
)

display(df)
```

38) DataFrame 다루기 - 연습문제

```python
# 1. score의 결측치를 중앙값으로 대체한 후 score의 합계를 구하시오.
import pandas as pd

df = pd.DataFrame({
    "score": [70, 80, 90, None, 60, 75, None, 85]
})

df["score"].fillna(df["score"].median(), inplace=True)
print("결측치 대체 후 score 합계:", df["score"].sum())
```

```python
# 2. score가 높은 순서로 정렬한 후 상위 30% 데이터를 추출하고, 그 데이터의 평균을 구하시오.
df_sorted = df.sort_values("score", ascending=False)
top_30_percent = df_sorted.head(int(len(df_sorted) * 0.3))
print("상위 30% 데이터의 평균:", top_30_percent["score"].mean())
```

```python
# 3. age가 30 이상 45 이하인 사람들의 income 평균을 구하시오.
df = pd.DataFrame({
    "age": [20, 25, 30, 35, 40, 45, 50],
    "income": [200, 250, 300, 350, 400, 450, 500]
})

df_filtered = df[(df["age"] >= 30) & (df["age"] <= 45)]
average_income = df_filtered["income"].mean()
print("age가 30 이상 45 이하인 사람들의 income 평균:", average_income)
```

```python
# 4. value에서 평균 + 2×표준편차보다 큰 값을 이상치라고 할 때, 이상치의 합을 구하시오.
df = pd.DataFrame({
    "value": [10, 12, 13, 15, 16, 17, 18, 20, 100]
})

df_mean = df["value"].mean()
df_std = df["value"].std()
outliers = df[df["value"] > df_mean + 2 * df_std]
print("이상치의 합:", outliers["value"].sum())
```

```python
# 5. IQR 방법을 이용하여 이상치를 제거한 후 score의 평균을 구하시오.
df = pd.DataFrame({
    "score": [60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200]
})
Q1 = df["score"].quantile(0.25)
Q3 = df["score"].quantile(0.75)
IQR = Q3 - Q1
df_cleaned = df[(df["score"] >= Q1 - 1.5 * IQR) & (df["score"] <= Q3 + 1.5 * IQR)]
print("IQR 방법을 이용하여 이상치를 제거한 후 score의 평균:", df_cleaned["score"].mean())
```

```python
# 6. 2025년 데이터만 추출하여 sales의 합계를 구하시오.
df = pd.DataFrame({
    "date": [
        "2024-01-10",
        "2024-03-20",
        "2025-01-15",
        "2025-05-10",
        "2025-07-20"
    ],
    "sales": [100, 200, 150, 300, 250]
})

df["date"] = pd.to_datetime(df["date"])
df_2025 = df[df["date"].dt.year == 2025]
print("2025년 데이터의 sales 합계:", df_2025["sales"].sum())
```

```python
# 7. code에서 숫자 부분만 추출하여 새로운 number 컬럼을 만들고, number의 평균을 구하시오.
df = pd.DataFrame({
    "code": [
        "A-100",
        "B-250",
        "A-300",
        "C-150"
    ]
})
df["number"] = df["code"].str.split("-").str[1].astype(int)
print("number의 평균:", df["number"].mean())
```

```python
# 8. 부서별 score 평균을 구한 후 평균이 두 번째로 높은 부서의 평균 점수를 구하시오.
df = pd.DataFrame({
    "department": ["A", "A", "B", "B", "C", "C"],
    "score": [70, 80, 90, 95, 60, 65]
})

department_avg = df.groupby("department")["score"].mean().reset_index()
department_avg_sorted = department_avg.sort_values("score", ascending=False)
second_highest_avg = department_avg_sorted.iloc[1]["score"]
print("평균이 두 번째로 높은 부서의 평균 점수:", second_highest_avg)
```

```python
# 9. 카테고리별 sales 합계를 구하고, 가장 큰 합계를 구하시오.
df = pd.DataFrame({
    "category": ["A", "A", "B", "B", "C", "C"],
    "sales": [100, 200, 300, 100, 150, 250]
})

df_category_sum = df.groupby("category")["sales"].sum().reset_index()
max_sales = df_category_sum["sales"].max()
print("가장 큰 카테고리별 sales 합계:", max_sales)
```

```python
# 10-1. date를 날짜형으로 변환하시오.
# 10-2. 연도-월별 sales 합계를 구하시오. 
# 10-3. 합계가 두 번째로 큰 연도-월의 sales 합계를 구하시오.
df = pd.DataFrame({
    "date": [
        "2024-01-10", "2024-01-20",
        "2024-02-10", "2024-02-20",
        "2024-03-10", "2024-03-20",
        "2024-04-10", "2024-04-20"
    ],
    "sales": [
        100, 150,
        200, 100,
        300, 250,
        150, 100
    ]
})

df["date"] = pd.to_datetime(df["date"])
df["year_month"] = df["date"].dt.to_period("M")
df_year_month_sum = df.groupby("year_month")["sales"].sum().reset_index()
df_year_month_sum_sorted = df_year_month_sum.sort_values("sales", ascending=False)
second_highest_sales = df_year_month_sum_sorted.iloc[1]["sales"]

print("합계가 두 번째로 큰 연도-월의 sales 합계:", second_highest_sales)
print("연도-월별 sales 합계:\n", df_year_month_sum)
print("합계가 두 번째로 큰 연도-월:\n", df_year_month_sum_sorted.iloc[1])
print("합계가 두 번째로 큰 연도-월의 sales 합계:", second_highest_sales)
```
