```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# 차트에 한글 깨지지 않도록 차트 폰트변경
plt.rcParams['font.family'] = 'Malgun Gothic'
# 차트에 "-"문자 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
# ----------------------------------------
# 반별 시험 평균 데이터 
# 원본 데이터셋에서 통계값 분석해서 아래 data를 추출해야 하지만
# 데이터를 추출하는 과정은 생략 
# ----------------------------------------

data = {
    "반": ["A반", "B반"],
    "중간고사": [78, 82],
    "기말고사": [84, 88]
}

avg_df = pd.DataFrame(data)

print(avg_df)

# ----------------------------------------
# 그룹 막대 그래프
# ----------------------------------------

avg_df.plot(
    x="반",
    y=["중간고사", "기말고사"],
    kind="bar"
)

plt.title("반별 시험 평균 비교")
plt.xlabel("반")
plt.ylabel("평균 점수")

# A반, B반을 가로로 표시
plt.xticks(rotation=0)

plt.show()
```
